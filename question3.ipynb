```python
import pandas as pd
```

## **1. Load the CSV file using Python.**

```python
df = pd.read_csv("orders_sample.csv", header=0)
df.info()
```

## **2. Remove exact duplicate rows.**

```python
df.drop_duplicates(inplace= True)
df.info()
```

## **3. Remove rows where quantity or unit_price is missing.**

```python
df.dropna(subset = ["quantity", "unit_price"], inplace = True)
df.info()
```

## **4.	Create column `line_revenue` = `quantity` * `unit_price`.**

```python
df["line_revenue"] = df["quantity"] * df["unit_price"]
df.info()
```

## **5.	Compute**

### **Casting to correct data type**

```python
df["order_id"] = df["order_id"].astype("string")
df["customer_id"] = df["customer_id"].astype("string")
df["product_id"] = df["product_id"].astype("string")
df["order_date"] = pd.to_datetime(df["order_date"])
df["quantity"] = df["quantity"].astype("Int64")
df["payment_status"] = df["payment_status"].astype("string")
df.info()
```

### **5.1 Total revenue by day**

```python
# Extract day from order date
df["day"] = df["order_date"].dt.day

# perform grouping and compute the total revenue of each day
df_total_revenue = df.groupby(["day"])["line_revenue"].sum()
```

### **5.2 total number of orders per customer**

```python
# group by customer id and count the number of order
df_total_order = df.groupby(["customer_id"])["order_id"].count()
```

### **5.3 payment success rate (paid / total orders)**

```python
# only collect record where payment_status is paid
df_paid = df[df["payment_status"] == "paid"]

# calculate the success rate by using the number of paid order divided by total number of order
success_rate = len(df_paid) / 22
```

## **6. Print The Final Results**

```python
print("Total revenue by day: ")
df_total_revenue
```

```python
print("Total number of orders per customers:")
df_total_order
```

```python
print(f"Payment success rate: {success_rate}")
```
